## Tarea 4: importar datos
Se descargó el conjunto Vertebral Column para reconocer sus formatos y cargar la versión de dos clases en pandas.

## 1. Bibliotecas y rutas

```python
from pathlib import Path
import hashlib, io, zipfile, requests
import pandas as pd
from scipy.io import arff
DATA = Path('../datos/vertebral')
DATA.mkdir(parents=True, exist_ok=True)
```

## 2. Descargar y comprobar el archivo

```python
url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/00212/vertebral_column_data.zip'
respuesta = requests.get(url, timeout=60)
respuesta.raise_for_status()
print('Estado HTTP:', respuesta.status_code)
archivo = zipfile.ZipFile(io.BytesIO(respuesta.content))
print('Archivos:', archivo.namelist())
archivo.extractall(DATA)
```

## 3. Cargar la versión de dos clases

```python
registros, metadatos = arff.loadarff(DATA / 'column_2C_weka.arff')
df = pd.DataFrame(registros)
print('Dimensiones:', df.shape)
df.info()
```

## 4. Validar la importación

```python
print(df.isna().sum())
print(df['class'].value_counts())
assert df.shape == (310, 7)
assert df['class'].nunique() == 2
print('Validación de importación: OK')
```

Se obtuvieron 310 observaciones, seis características numéricas y una etiqueta, sin valores nulos.
